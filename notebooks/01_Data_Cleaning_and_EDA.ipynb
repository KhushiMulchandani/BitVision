## INTRODUCTION
Bitcoin is one of the world's most volatile and widely traded cryptocurrencies, making accurate price forecasting a challenging yet valuable task for investors, researchers, and financial institutions. Understanding historical price movements and identifying market patterns through data analysis can significantly improve forecasting models and investment decision-making.

Rather than functioning as a simple prediction tool, <b>BitVision</b> aims to provide a complete analytics pipeline that includes data preprocessing, exploratory analysis, feature engineering, model training, prediction services, and interactive visualizations.

This notebook represents Version 1.0 (Foundation) of the <b>BitVision</b> platform. The insights obtained during this stage will serve as the foundation for feature engineering and predictive model development in subsequent versions.

The cryptocurrency market is characterized by rapid price fluctuations, high volatility, and continuously changing market dynamics. These characteristics make price prediction a complex task requiring robust data preprocessing and advanced analytical techniques.

The primary business challenge is to transform raw historical market data into meaningful insights that can support reliable forecasting models. Poor-quality data, missing values, inconsistent records, and insufficient feature representation can significantly reduce prediction accuracy.

To address this challenge, <b>BitVision</b> follows a structured data science pipeline. Ultimately, the platform aims to provide an intelligent analytics system that enables users to visualize historical trends, understand market behavior, and generate data-driven price predictions.

## EXPLORATORY DATA ANALYSIS

```python
plt.style.use("ggplot")

plt.rcParams["figure.figsize"] = (14,6)
plt.rcParams["font.size"] = 11
plt.rcParams["axes.titlesize"] = 15
plt.rcParams["axes.labelsize"] = 12
```

### 1. Closing Price Trend
The closing price is one of the most significant indicators in financial markets, as it represents the final trading price for a particular day. Visualizing the closing price over time helps identify long-term trends, periods of rapid growth, market crashes, and overall price volatility.

```python
plt.figure(figsize=(15,6))

plt.plot(df["Date"], df["Close"])

plt.title("Bitcoin Closing Price Over Time")
plt.xlabel("Date")
plt.ylabel("Closing Price (USD)")

plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

#### OBSERVATION:

The Bitcoin closing price remained relatively low during the initial years, followed by multiple periods of rapid appreciation and correction. The most significant upward trend occurred during the later years of the dataset, demonstrating substantial long-term price growth accompanied by high volatility.

#### BUSINESS INSIGHTS:

Bitcoin has experienced multiple bullish and bearish market cycles rather than steady linear growth. Investors should therefore expect significant price fluctuations while recognizing the asset's long-term appreciation potential.

### 2. Trading Volume Trend
Trading volume indicates the total quantity of Bitcoin traded during a day. Analyzing trading volume helps understand market participation and often highlights periods of increased investor activity.

```python
plt.figure(figsize=(15,6))

plt.plot(df["Date"], df["Volume"])

plt.title("Bitcoin Trading Volume Over Time")
plt.xlabel("Date")
plt.ylabel("Volume")

plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

#### OBSERAVTION:

Trading volume remained comparatively low during the early years but increased substantially in later years. Several prominent spikes indicate periods of exceptionally high trading activity.

#### BUSINESS INSIGHTS:

Higher trading volume generally reflects increased investor participation and heightened market interest, particularly during periods of major price movements.

### 3. Distribution of Numerical Features
Understanding the distribution of numerical variables helps identify skewness, concentration of values, and the overall spread of the dataset.

```python
numerical_columns = ["Open", "High", "Low", "Close", "Volume"]

for column in numerical_columns:
    plt.figure(figsize=(8,4))

    sns.histplot(
        df[column],
        kde=True,
        bins=40
    )
    
    plt.xlabel(column)
    plt.ylabel("Frequency")

    plt.title(f"Distribution of {column}")

    plt.tight_layout()
    plt.show()
```

#### OBSERVATION:

##### <i style="color:blue;"> Distribution of Opening Prices: </i>
The distribution of opening prices is highly right-skewed, with most observations concentrated at lower price levels and fewer observations occurring at very high prices.

##### <i style="color:blue;">High Price Distribution:</i>
The High price distribution exhibits a strong positive skew similar to the Open price distribution, with a long tail extending toward larger price values.

##### <i style="color:blue;">Low Price Distribution:</i>
The Low price values are also positively skewed, indicating that lower price ranges dominate the historical dataset while higher values occur less frequently.

##### <i style="color:blue;">Closing Price Distribution:</i>
Closing prices demonstrate a positively skewed distribution, with the majority of observations clustered at lower values and relatively fewer observations at higher prices.

##### <i style="color:blue;">Volume Distribution:</i>
Trading volume displays an extremely right-skewed distribution, with a few days recording exceptionally high trading activity.

#### BUSINESS INSIGHTS:

##### <i style="color:blue;">Distribution of Opening Prices:</i>
Bitcoin spent a considerable portion of its history trading at relatively low prices before experiencing substantial long-term appreciation.

##### <i style="color:blue;">High Price Distribution:</i>
Extreme high prices occurred infrequently but represent genuine market rallies rather than anomalies.

##### <i style="color:blue;">Low Price Distribution:</i>
The distribution reflects Bitcoin's gradual market maturation over time.

##### <i style="color:blue;">Closing Price Distribution:</i>
The distribution highlights Bitcoin's long-term upward trajectory rather than a stable price range.

##### <i style="color:blue;">Volume Distribution:</i>
Large spikes in trading activity often correspond to periods of heightened market volatility and increased investor participation.

### 4. Boxplots for Outlier Detection
Bitcoin naturally experiences extreme price movements. Outliers are part of the market, not necessarily errors.

```python
for column in numerical_columns:
    plt.figure(figsize=(8,3))

    sns.boxplot(x=df[column])

    plt.title(f"Boxplot of {column}")

    plt.tight_layout()
    plt.show()
```

#### OBSERVATION:

##### <i style="color:blue;"> Distribution of Opening Prices: </i>
Several observations appear beyond the upper whisker, indicating numerous high-value observations.

##### <i style="color:blue;">High Price Distribution:</i>
The High price also contains multiple upper-end outliers, reflecting days with exceptionally high trading prices.

##### <i style="color:blue;">Low Price Distribution:</i>
The Low price distribution contains several upper-end outliers while maintaining a relatively compact interquartile range.

##### <i style="color:blue;">Closing Price Distribution:</i>
The Close price distribution exhibits multiple high-value observations beyond the whiskers.

##### <i style="color:blue;">Volume Distribution:</i>
Trading volume contains several extreme observations, with a few days recording exceptionally high market activity.

#### BUSINESS INSIGHTS:

##### <i style="color:blue;">Distribution of Opening Prices:</i>
These values correspond to genuine market behavior resulting from Bitcoin's rapid appreciation and therefore should not be treated as erroneous outliers.

##### <i style="color:blue;">High Price Distribution:</i>
Such observations are expected in cryptocurrency markets and represent authentic market dynamics.

##### <i style="color:blue;">Low Price Distribution:</i>
These observations indicate periods during which the overall market price shifted to substantially higher valuation levels.

##### <i style="color:blue;">Closing Price Distribution:</i>
Rather than data quality issues, these represent Bitcoin's historical bull markets and were intentionally retained.

##### <i style="color:blue;">Volume Distribution:</i>
These spikes often coincide with important market events and therefore provide valuable information for predictive modeling.

### 5. Correlation Analysis
A correlation matrix helps identify linear relationships between numerical variables.

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    df[numerical_columns].corr(),
    annot=True,
    cmap="RdYlBu_r",
    linewidths=.5,
    square=True,
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.show()
```

#### OBSERVATION:

Open, High, Low, and Close prices exhibit an almost perfect positive correlation, while Volume shows only a moderate positive correlation with the price variables.

#### BUSINESS INSIGHTS:

Price-related variables move together because they describe different aspects of the same trading session. Volume provides additional information regarding market participation rather than directly determining price.

### 6. Pairwise Relationships

```python
sns.pairplot(
    df[numerical_columns],
    diag_kind="kde",
    corner=True
)

plt.tight_layout()
plt.show()
```

#### OBSERVATION:

Strong linear relationships are observed among the price variables, whereas the relationship between Volume and price variables appears comparatively weaker and more dispersed.

#### BUSINESS INSIGHTS:

The pairwise relationships confirm that price variables are highly interdependent, while trading volume contributes complementary information that may improve predictive models.

### 7. Monthly Closing Price Trend

```python
df["Month"] = df["Date"].dt.month_name()
```

```python
plt.figure(figsize=(12,6))

sns.boxplot(
    x="Month",
    y="Close",
    data=df,
    order=[
        "January","February","March","April","May","June",
        "July","August","September","October","November","December"
    ]
)

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

#### OBSERVATION:

The monthly boxplots reveal considerable variability across all months, with no consistent seasonal pattern in Bitcoin closing prices.

#### BUSINESS INSIGHTS:

Unlike many traditional financial assets, Bitcoin does not demonstrate strong month-based seasonality, suggesting that market dynamics are driven more by broader economic and investor factors than calendar effects.

### 8. Year-wise Closing Price Trend

```python
df["Year"] = df["Date"].dt.year
```

```python
plt.figure(figsize=(12,6))

sns.lineplot(
    x="Year",
    y="Close",
    data=df,
    estimator="mean",
    marker="o"
)

plt.title("Average Bitcoin Closing Price by Year")

plt.tight_layout()
plt.show()
```

#### OBSERVATION:

The average annual closing price demonstrates a clear long-term upward trend, despite temporary declines during certain years.

#### BUSINESS INSIGHTS:

The yearly trend reinforces Bitcoin's long-term growth trajectory while highlighting the presence of cyclical market corrections.
